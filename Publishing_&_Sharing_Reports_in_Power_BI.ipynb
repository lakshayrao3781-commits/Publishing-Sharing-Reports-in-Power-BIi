Q1. Explain why publishing and sharing reports is considered more important than just building dashboards
in Power BI Desktop.

Building reports locally in Power BI Desktop only creates isolated analytics on a single developer machine. Publishing and sharing turns static models into actionable business intelligence by enabling cross-team collaboration, scheduled automatic data refreshes, central security controls, and real-time decision-making for stakeholders across web and mobile platforms.

Q2. Define Power BI Service and explain its role in comparison to Power BI Desktop in the reporting lifecycle.

Power BI Desktop: A Windows desktop authoring tool used for data connection, data transformation (Power Query), modeling (DAX), and visual report creation.   Power BI Service: A cloud-based SaaS platform where reports created in Desktop are published. Its role is content distribution, dashboard creation, automated scheduled refresh, security management (RLS, workspace permissions), and enterprise collaboration.   

Q3. What does publishing a report mean in Power BI? List any two outcomes that occur after a report is
published.

Publishing a report means uploading the .pbix file (including dataset, schema, DAX measures, and visuals) from Power BI Desktop to a designated workspace in Power BI Service.  

 Two outcomes after publishing:Cloud Accessibility: End-users can view and interact with the report via web browsers or mobile apps without needing Power BI Desktop installed.   

 Dataset Separation: The published dataset is decoupled from the report layout, allowing it to be scheduled for automatic refresh or reused by other reports.   

Q4. List and explain the steps involved in publishing a report from Power BI Desktop to Power BI Service in the
correct sequence.

Save the Report: Ensure all changes in Power BI Desktop are saved locally as a .pbix file.  

 Sign In: Click Sign In at the top right of Power BI Desktop using your organizational account.   

 Trigger Publish: Go to the Home tab and click Publish.   Select Destination Workspace: Choose the specific Power BI Service workspace (e.g., My Workspace or a shared team workspace) and click Select.   

 Confirmation & Verification: Wait for the "Success!" message and click the provided link to open the report directly in Power BI Service to verify visuals and refresh settings.

Q6. Explain the difference between the following three sharing methods in Power BI Service:
Direct Report Sharing
Workspace Access
Power BI Apps
Also mention one use case for each.

Direct Report Sharing:Mechanism: Granting specific individuals view or edit access to a single report via their email address.   Use Case: Sharing a quick ad-hoc financial summary with an external consultant or a single executive manager.   Workspace Access:Mechanism: Adding users directly to a workspace with specific roles (Admin, Member, Contributor, Viewer).   Use Case: Team collaboration among business analysts and developers co-authoring and testing reports.   Power BI Apps:Mechanism: Packaging a curated collection of dashboards, reports, and datasets from a workspace and deploying them to a large user group with a unified interface.   Use Case: Broad, enterprise-wide distribution of standardized quarterly sales dashboards to hundreds of regional managers.   

Q7. Describe the four workspace roles available in Power BI and explain how access control is managed using
these roles.

Admin: Full control over the workspace, including managing user permissions, deleting the workspace, and modifying app settings.  

 Member: Can add users with lower permissions, publish/edit content, and update published Apps.  

  Contributor: Can create, edit, and delete reports/datasets inside the workspace, but cannot manage permissions or publish Apps.
  
  Viewer: Read-only access to view and interact with reports/dashboards in the workspace without edit rights.  

  How access control is managed: Permissions follow the principle of least privilege. Developers are assigned Contributor or Member roles for content creation, while consumers are assigned Viewer roles or consume content through published Apps, preventing unauthorized edits or data leaks.

Q8. Why are Power BI Apps considered more secure and scalable than direct sharing?
Explain this from a governance and enterprise usage perspective.

Governance & Version Control: Apps create a staging environment. Developers can make ongoing modifications in the development workspace without affecting end-users until the App is explicitly republished.  

 Scalability: Managing access for thousands of users via Microsoft 365 security groups inside an App is far cleaner than maintaining individual email permissions on dozens of standalone reports.  

  Centralized Security: Prevents accidental modification or deletion of source files and datasets by consumers while giving admins centralized auditing capabilities over content distribution.  

Q9. Explain Row-Level Security (RLS) in Power BI.
How does RLS ensure that different users see different data using the same report?

Row-Level Security (RLS) is a security feature used to restrict data access at the database/dataset row level based on user identity or role membership.  

 How RLS works with the same report:

 Roles are defined in Power BI Desktop using DAX expression filters on dimension tables (e.g., [Region] = USERPRINCIPALNAME()).   

 Users are assigned to these roles in Power BI Service.  

  When two users open the exact same report, Power BI applies the DAX filter dynamically at query execution time. For example, the North Region Manager only sees rows where Region = "North", while the South Region Manager sees rows where Region = "South".

Q10.Assume an organization shares reports without using Apps, workspace roles, or RLS.
Discuss any three risks or issues that may arise related to security, governance, or data misuse, and explain
how Power BI best practices help prevent them.

Risk 1: Sensitive Data Exposure / Data BreachesIssue: Without RLS or workspace roles, unauthorized employees can see sensitive financial metrics or salary data intended for executives.   Best Practice: Implement Row-Level Security (RLS) and assign read-only Viewer roles or packaged Apps to enforce strict data visibility limits.  

 Risk 2: Accidental Overwriting or Deletion of Production ReportsIssue: Directly sharing workspace access with write permissions allows non-technical users to modify formulas or delete reports unintentionally.   Best Practice: Separate environments by keeping developers in the workspace as Contributors and serving consumers through published Power BI Apps.  

 Risk 3: Unmanaged Version Drift and Shared Links SprawlIssue: Sharing one-off direct links creates multiple inconsistent copies of data across teams, breaking the "single source of truth".   Best Practice: Enforce App Publishing combined with Organizational Workspaces and security groups, ensuring all users consume approved, auto-refreshed enterprise reports from a single governed location.
